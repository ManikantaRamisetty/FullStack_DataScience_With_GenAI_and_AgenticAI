# as it is numeric need to apply mean median mode logic to fill missing data

```python
import numpy as np
```

```python
emp_cleanData['Age'] = emp_cleanData['Age'].fillna(np.mean(pd.to_numeric(emp_cleanData['Age'])))
```

```python
emp_cleanData['Age'] # applied mean startegy (avg of all numeric values)
```

```python
emp_cleanData['Location']
```

```python
emp_cleanData['Location'] = emp_cleanData['Location'].fillna(np.mode(pd.to_numeric(emp_cleanData['Location'])))
```

```python
emp_cleanData['Location'] = emp_cleanData['Location'].fillna(emp_cleanData['Location'].mode()[0])
```

```python
emp_cleanData['Location']
```

```python
emp_cleanData
```

```python
emp_cleanData['Exp'] = emp_cleanData['Exp'].fillna(np.mean(pd.to_numeric(emp_cleanData['Exp'])))
```

```python
#all types are object converting int & string (catogerical)
emp_cleanData['Age'] = emp_cleanData['Age'].astype(int)
emp_cleanData['Exp'] = emp_cleanData['Exp'].astype(int)
emp_cleanData['Salary'] = emp_cleanData['Salary'].astype(int)



emp_cleanData['Name'] = emp_cleanData['Name'].astype(int)
emp_cleanData['Domain'] = emp_cleanData['Domain'].astype(int)
emp_cleanData['Location'] = emp_cleanData['Location'].astype(int)
```

```python
#all types are object converting int & string (catogerical)
# astype convert the sys defined types(object) to userdefined datatype
emp_cleanData['Age'] = emp_cleanData['Age'].astype(int)
emp_cleanData['Exp'] = emp_cleanData['Exp'].astype(int)
emp_cleanData['Salary'] = emp_cleanData['Salary'].astype(float)



emp_cleanData['Name'] = emp_cleanData['Name'].astype('category')
emp_cleanData['Domain'] = emp_cleanData['Domain'].astype('category')
emp_cleanData['Location'] = emp_cleanData['Location'].astype('category')
```
